```python
import pandas as pd

df = pd.read_csv(
    "../data/spam.csv",
    encoding="latin-1"
)

df = df.iloc[:, :2]

df.columns = [
    "label",
    "message"
]
```

```python
df.head()

df.info()

df["label"].value_counts()

df["message_length"] = df["message"].apply(len)

df.groupby(
    "label"
)["message_length"].mean()
```
